```python
num = 29
is_prime = True

# Edge case: Numbers 1 and below are not prime by definition
if num <= 1:
    is_prime = False
else:
    # We check all numbers from 2 up to (but not including) the number itself
    for i in range(2, num):
        # If 'num' divides evenly by any of these numbers, it is not prime
        if num % i == 0:
            is_prime = False
            break  # We found a divisor, no need to keep checking!

if is_prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
def compress_string(text):
    # 1. Handle the edge case of an empty string
    if text == "":
        return ""
        
    compressed = ""
    
    # 2. Set up our initial tracking variables
    current_char = text[0]
    count = 0
    
    # 3. Loop through every character in the text
    for char in text:
        if char == current_char:
            # Still on the same run, just add 1 to the count
            count += 1
        else:
            # The character changed! 
            # Add the OLD character and its count to our result
            compressed += f"{current_char}{count}"
            
            # Reset our trackers for the NEW character
            current_char = char
            count = 1
            
    # 4. The Loop's End (The tricky part)
    # The loop finishes, but the final character run hasn't been added yet!
    compressed += f"{current_char}{count}"
    
    return compressed

# Tests
print(compress_string("AAABBC"))      
print(compress_string("WWWWWWWWWWW")) 
print(compress_string("ABBA"))        
print(compress_string(""))
```

```python
def update_inventory(current_stock, shipment_list):
    for item in shipment_list:
        # Check if we already carry this item
        if item in current_stock:
            current_stock[item] += 1
        # Brand new item!
        else:
            current_stock[item] = 1
            
    return current_stock

# Testing
stock = {"apples": 5, "bananas": 2}
delivery = ["apples", "oranges", "apples", "kiwi"]
print(update_inventory(stock, delivery))
```

```python
def group_by_length(words):
    grouped = {}
    for word in words:
        lenght = len(word)
        if lenght in grouped:
            grouped[lenght].append(word)
        else:
            grouped[lenght]=[word]
    return grouped

vocab = ["cat", "dog", "bear", "frog", "ox"]
print(group_by_length(vocab))
```

```python
def find_winner(votes):
    vote_counts = {}
    for candidate in votes:
        vote_counts[candidate]=vote_counts.get(candidate, 0) + 1

    winner = ""
    highest_votes = 0
    for candidate , count in vote_counts.items():
        if count > highest_votes:
            highest_votes = count
            winner = candidate
    return winner
ballots = ["Alice", "Bob", "Alice", "Charlie", "Bob", "Alice"]
print(find_winner(ballots))
        
```

```python
def process_bookings(flights, requests):
    failed_bookings = []
    
    for req in requests:
        
        req_name = req["name"]
        req_flight = req["flight"]
        
       
        if req_flight in flights:
            
            
            flight_data = flights[req_flight]

            if len(flight_data["passengers"]) < flight_data["capacity"]:
      
                flight_data["passengers"].append(req_name)
            else:
                
                failed_bookings.append(req_name)
                
        else:
            # Fail: Flight key does not exist
            failed_bookings.append(req_name)
            
    return failed_bookings

# --- Testing ---
flights_db = {
    "NY101": {"destination": "New York", "capacity": 2, "passengers": ["Alice"]},
    "LA202": {"destination": "Los Angeles", "capacity": 1, "passengers": []}
}

new_requests = [
    {"name": "Bob", "flight": "NY101"},      
    {"name": "Charlie", "flight": "NY101"},  
    {"name": "Diana", "flight": "LA202"},    
    {"name": "Evan", "flight": "MIA303"}     
]

print(process_bookings(flights_db, new_requests))
```

```python
def is_palindrome(text):
   
    cleaned_text = ""
    for char in text:
        if char.isalnum():
            cleaned_text += char.upper()
    return cleaned_text == cleaned_text[::-1]
            

    
# Testing
print(is_palindrome("A man a plan a canal Panama")) # Expected: True
print(is_palindrome("hello"))                       # Expected: False
```

```python
def fizz_buzz(n):
    results = []
    for i in range(1, n+1):
        if i % 3 == 0 and i % 5 == 0:
            results.append("fuzzbuzz")
        elif i % 3 == 0:
            results.append("fizz")
        elif i % 5 == 0:
            results.append("buzz")
        else:
            results.append(str(i))
    return results
print(fizz_buzz(12))
            
```

```python
def are_anagrams(str1, str2):
    # Quick check: if lengths are different, they cannot be anagrams
    if len(str1) != len(str2):
        return False
        
    counts1 = {}
    counts2 = {}
    for char in str1:
        counts1[char] = counts1.get(char,0)+1
    for char in str2:
        counts2[char]=counts2.get(char,0)+1
        
    
    return counts1 == counts2

# Testing
print(are_anagrams("listen", "silent")) # Expected: True
print(are_anagrams("apple", "pale"))    # Expected: False
```

```python
def top_k_frequent(nums, k):
   
    counts = {}
    for num in nums:
        counts[num] = counts.get(num, 0) + 1
        
    
    sorted_nums = sorted(counts.keys(), key=lambda x: counts[x], reverse=True)
    
    # Step 3: Return only the first 'k' elements using a slice
    return sorted_nums[:k]


print(top_k_frequent([1, 1, 1, 2, 2, 3,10,9,7,6], 3)) 
# Expected Output: [1, 2]
```

```python
def flatten_matrix(matrix):
    flat_list = []
    
   
    for row in matrix:
        
      
        for item in row:
            flat_list.append(item)
            
    return flat_list

# Testing
print(flatten_matrix([[1, 2], [3, 4], [5, 6]])) 
# Expected Output: [1, 2, 3, 4, 5, 6]
```

```python
def merge_sales(transactions):
    totals = {}
    for record in transactions:
        rep_name = record['rep']
        sale_amount = record['amount']
        totals[rep_name]=totals.get(rep_name,0)+ sale_amount
               
    return totals

# Testing
logs = [
    {"rep": "Alice", "amount": 100}, 
    {"rep": "Bob", "amount": 50}, 
    {"rep": "Alice", "amount": 75},
    {"rep": "Bob", "amount": 60},
]
print(merge_sales(logs)) 
# Expected Output: {'Alice': 175, 'Bob': 50}
```

```python
def rebalance_stock(warehouses, target_item, min_threshold):
    low_stock_locations = []
    
   
    for wh_id, inventory in warehouses.items():
        
 
        current_stock = inventory.get(target_item, 0)
        
       
        if current_stock < min_threshold:
            low_stock_locations.append(wh_id)
            
    return low_stock_locations


db = {
    "WH_A": {"laptops": 15, "monitors": 2},
    "WH_B": {"laptops": 3, "monitors": 12},
    "WH_C": {"laptops": 0, "monitors": 8},
}
print(rebalance_stock(db, "laptops", 5))
```
